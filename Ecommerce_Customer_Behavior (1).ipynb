* **E-Commerce Customer Behavior Analysis**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

* **Data Loading**

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/diamonds.csv"
df = pd.read_csv(url)

df.head()
df.shape
```

* **Data Understanding**

```python
df.info()
```

```python
df.describe()
```

* **Visualization**

```python
plt.figure(figsize=(6,4))
sns.countplot(x='cut', data=df)
plt.title("Product Category Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['price'], kde=True)
plt.title("Price Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='cut', y='price', data=df)
plt.title("Price by Category")
plt.show()
```

* **Insights**

  Customers tend to prefer certain product categories more, and pricing plays a key role in purchasing behavior across categories.

* **Conclusion**

  This project analyzed customer behavior in an e-commerce setting.
  The analysis helped identify product trends, pricing patterns, and category preferences, which can help businesses improve sales strategies.
